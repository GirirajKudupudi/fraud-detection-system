```python
import pandas as pd


import os
print(os.listdir('../data/raw/'))
```

```python
df = pd.read_csv('../data/raw/synthetic_indian_upi_fraud_data.csv')
print(f"Shape: {df.shape}")
print(f"\nColumns:\n{df.columns.tolist()}")
df.head()
```

```python
df.info()
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python

print("Fraud distribution:")
print(df['is_fraud'].value_counts())
print(f"\nFraud percentage: {df['is_fraud'].mean()*100:.2f}%")
```

```python
print(df.columns.tolist())
```

```python
print("Fraud distribution:")
print(df['FraudFlag'].value_counts())
print(f"\nFraud percentage: {df['FraudFlag'].mean()*100:.2f}%")
```

```python
print("Transaction Types:")
print(df['TransactionType'].value_counts())
print(f"\nMerchant Categories:")
print(df['MerchantCategory'].value_counts())
print(f"\nBank Names:")
print(df['BankName'].value_counts())
```

```python
print("Amount stats:")
print(df['Amount'].describe())
print(f"\nFraud vs Non-Fraud amounts:")
print(df.groupby('FraudFlag')['Amount'].describe())
```

```python
print("Risk signals in fraud vs non-fraud:")
for col in ['UnusualLocation', 'UnusualAmount', 'NewDevice', 'FailedAttempts']:
    print(f"\n{col}:")
    print(df.groupby('FraudFlag')[col].mean())
```

```python
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
df['hour'] = df['Timestamp'].dt.hour
df['day_of_week'] = df['Timestamp'].dt.day_name()

print("Fraud by hour:")
fraud_by_hour = df.groupby('hour')['FraudFlag'].mean().sort_values(ascending=False)
print(fraud_by_hour.head(10))

print("\nFraud by day:")
print(df.groupby('day_of_week')['FraudFlag'].mean().sort_values(ascending=False))
```

```python
print("Fraud rate by merchant category:")
print(df.groupby('MerchantCategory')['FraudFlag'].mean().sort_values(ascending=False))

print("\nFraud rate by transaction type:")
print(df.groupby('TransactionType')['FraudFlag'].mean().sort_values(ascending=False))

print("\nFraud rate by bank:")
print(df.groupby('BankName')['FraudFlag'].mean().sort_values(ascending=False))
```

```python
print("Location stats for fraud vs non-fraud:")
for col in ['Latitude', 'Longitude']:
    print(f"\n{col}:")
    print(df.groupby('FraudFlag')[col].describe())
```

```python
print(df['TransactionFrequency'].head(10))
print(f"\nType: {df['TransactionFrequency'].dtype}")
```

```python
df3 = pd.read_csv('../data/raw/bank_transactions_data_2.csv')
print(f"Shape: {df3.shape}")
print(f"\nColumns: {df3.columns.tolist()}")

# Find fraud column
for col in df3.columns:
    if 'fraud' in col.lower() or 'flag' in col.lower() or 'label' in col.lower() or 'anom' in col.lower():
        print(f"\n{col}:")
        print(df3[col].value_counts())
        print(f"Fraud rate: {df3[col].mean()*100:.2f}%")

# THE KEY TEST - check correlations
print("\nCorrelations with fraud:")
corr = df3.corr(numeric_only=True).iloc[:, -1].abs().sort_values(ascending=False)
print(corr.head(15))
```
